### Step 1 — Client Scenario: MNM AG Vendor Management

MNM AG is a fictional company using Microsoft Dynamics 365 Finance & Operations to manage vendors across multiple legal entities. The company wants to evaluate whether its AI assistant provides accurate, complete and reliable answers to vendor management questions.

The evaluation will use LLM-as-a-Judge to assess answers against company reference information, with particular attention to incorrect claims, unsupported information and potentially misleading guidance.
